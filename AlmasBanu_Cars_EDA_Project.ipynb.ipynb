```python
# Fill Missing Values
df["Engine HP"]=df["Engine HP"].fillna(df["Engine HP"].mean())
df["Engine Cylinders"]=df["Engine Cylinders"].fillna(df["Engine Cylinders"].mean())
df["Number of Doors"]=df["Number of Doors"].fillna(df["Number of Doors"].mean())
df["Engine Fuel Type"]=df["Engine Fuel Type"].fillna(df["Engine Fuel Type"].mode()[0])
df["Market Category"]=df["Market Category"].fillna(df["Market Category"].mode()[0])
```

```python
df.isnull().sum()
```

```python
#Visualization
#1.Which brands are most common?
sns.countplot(y="Make",data=df,order=df["Make"].value_counts().index[:10])
plt.title("Top 10 Most Common Car Brands")
plt.xlabel("Count")
plt.ylabel("Car Brand")
plt.show()
```

Ford, Chevrolet, and Volkswagen have the highest number of cars in the dataset, while a few brands appear less frequently.

```python
#2.Which fuel type is most common?
sns.countplot(data=df,y="Engine Fuel Type")
plt.title("Engine Fuel Type Distribution")
plt.xlabel("Count")
plt.ylabel("Engine Fuel Type")
plt.show()
```

Most cars use regular unleaded fuel, showing that conventional fuel vehicles are more common than alternative fuel vehicles.

```python
#3.How are car prices distributed?
sns.histplot( df["MSRP"],bins=30, kde = True )
plt.title("Distribution of Car Prices")
plt.xlabel("MSRP")
plt.ylabel("Frequency")
plt.show()
```

Most cars are moderately priced, while only a few expensive luxury cars appear as high-price values.

```python
#4.How does vehicle age relate to price?
df["Vehicle Age"]= 2026 - df["Year"]
sns.scatterplot(data=df,x="Vehicle Age",y="MSRP")
plt.title("MSRP vs Vehicle Age")
plt.xlabel("Vehicle Age")
plt.ylabel("MSRP")
plt.show()
```

In general, newer cars tend to have higher prices, although there are some exceptions.

```python
#5.Does mileage influence price?
sns.scatterplot(data=df,x="highway MPG",y="MSRP")
plt.title("Highway MPG vs Car Price")
plt.xlabel("Highway MPG")
plt.ylabel("MRSP")
plt.figure(figsize=(10,6))
plt.show()
```

Fuel efficiency does not show a strong relationship with car price, indicating that other factors also affect pricing.

```python
#6.Do automatic and manual cars differ in price?
sns.boxplot(x="Transmission Type",
            y="MSRP",
            data=df)
plt.title("Transmission Type vs Car Price")
plt.xlabel("Transmission Type")
plt.ylabel("MSRP")
plt.xticks(fontsize=7)
plt.figure(figsize=(10,6))
plt.show()
```

Automatic cars generally have higher prices than manual cars, with a few expensive outliers.

```python
#Correlation matrix values before the heatmap
numeric_df = df.select_dtypes(include=np.number)
corr = numeric_df.corr()
print(corr)
```

```python
#7.Which features are strongly correlated?
numeric_df = df.select_dtypes(include=np.number)
corr = numeric_df.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

Some numerical features are positively related to price, while others show only a weak relationship.

```python
#8.Where are the important outliers?
plt.figure(figsize=(8,5))
sns.boxplot(y="MSRP",data=df)
plt.title("Outliers in Car Prices")
plt.ylabel("MSRP")
plt.show()
```

The graph shows a few high-priced outliers, which are likely luxury or premium vehicles.
